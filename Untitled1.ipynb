```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
diabetes_dataset = pd.read_csv('diabetes.csv')
```

```python
columns_with_missing = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]

diabetes_dataset[columns_with_missing] = (
    diabetes_dataset[columns_with_missing]
    .replace(0, np.nan)
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('svm', SVC(kernel='linear'))
])
```

```python
diabetes_dataset.groupby('Outcome').mean()
```

**Model Training**

```python

X = diabetes_dataset.drop(columns='Outcome')
Y = diabetes_dataset['Outcome']

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    stratify=Y,
    random_state=2
)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python


models = {
    "Linear SVM": svm.SVC(kernel='linear'),
    "RBF SVM": svm.SVC(kernel='rbf'),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Random Forest": RandomForestClassifier(random_state=2),
    "KNN": KNeighborsClassifier()
}

for name, model in models.items():
    model.fit(X_train, Y_train)
    prediction = model.predict(X_test)
    accuracy = accuracy_score(Y_test, prediction)
    print(f"{name}: {accuracy:.4f}")
```

**RBF**

```python
pipeline_rbf = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(kernel='rbf'))
])

parameters_rbf = {
    'svm__C': [0.01, 0.1, 1, 10, 100],
    'svm__gamma': ['scale', 0.001, 0.01, 0.1, 1]
}

grid_rbf = GridSearchCV(
    pipeline_rbf,
    parameters_rbf,
    cv=10,
    scoring='accuracy',
    n_jobs=-1
)

grid_rbf.fit(X_train, Y_train)

print("Best RBF parameters:", grid_rbf.best_params_)
print("Best RBF CV accuracy:", grid_rbf.best_score_)
```

```python
rbf_model = grid_rbf.best_estimator_

rbf_prediction = rbf_model.predict(X_test)

print(
    "RBF Test accuracy:",
    accuracy_score(Y_test, rbf_prediction)
)
```
